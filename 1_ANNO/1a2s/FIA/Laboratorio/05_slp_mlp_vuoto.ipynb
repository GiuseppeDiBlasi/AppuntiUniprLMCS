Implementare un neurone artificiale che prenda in input $N = 8$ variabili:
- Vedere il modello come una funzione $f: \mathbb{R}^N \rightarrow \mathbb{R}$ e definirla
- Definire la funzione di attivazione di Heaviside
- Definire la funzione di attivazione sigmoide con $\beta = 1$
- Creare un neurone che usa la funzione di Heaviside come attivazione e un altro che usa la sigmoide.

```python
import numpy as np

def heaviside(x):
    if x > 0:
        return 1
    else:
        return 0
    
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def pre_activation(x, w, b):
    return np.dot(x, w) + b

x = np.random.randn(1, 8)
w = np.random.randn(8, 1)
b = np.random.rand(1)

y = heaviside(pre_activation(x, w, b))

print(y)
```
