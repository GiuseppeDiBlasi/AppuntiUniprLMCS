# Le funzioni $F: \mathbf{R}^N \rightarrow \mathbf{R}^M$

## Esempio: coordinate sferiche

Sia $F: \mathbf{R}^2 \rightarrow \mathbf{R}^3$ definita nel seguente modo:

$$
F(\theta, \phi) =
\begin{cases}
y_1 &= \rho \sin(\theta) \cos (\phi)\\
y_2 &= \rho \sin(\theta) \sin(\phi)\\
y_3 &= \rho \cos(\theta)
\end{cases}
$$

con $\rho$ fissato (ad es. $\rho = 1$), $0 \leq \theta \leq \pi$ e $0 \leq \phi \leq 2 \pi$

```python
import numpy as np
import matplotlib.pyplot as plt

RHO = 1
```

$f_1 (\theta, \phi) = \rho \sin(\theta) \cos(\phi)$

```python
def f_1(theta, phi):
    phi* np.sin(theta) * np.cos(phi)
```

$f_2(\theta, \phi) = \rho \sin(\theta) \sin(\phi)$

```python
def f_2(theta, phi):
    phi* np.sin(theta) * np.sin(phi)
```

$f_3(\theta, \phi) = \rho \cos(\theta)$

```python
def f_3(theta, phi):
    RHO * np.cos(theta)
```

```python
def F(theta, phi): 
    return [f_1(x, y), f_2(x, y), f_3(x, y)]
```

I nostri vettori di input saranno $\underline{x} = (\theta, \phi)$ in $\mathbf{R}^2$.

Creiamo una meshgrid:

```python
lin_x = np.linspace(0, np.pi, 100)
lin_y = np.linspace(0, 2 * np.pi, 10)

x, y = np.meshgrid(lin_x, lin_y)

print(lin_x.shape)
print(lin_y.shape)
print(x.shape)
print(y.shape)
```

## Esempio: $f$ continua separatamente in $x$ e $y$

Sia $f: \mathbf{R}^2 \rightarrow \mathbf{R}$, definita nel seguente modo:

$$
f(x, y) =
\begin{cases}
\frac{xy}{x^2 + y^2} & \text{se } (x, y) \not = (0, 0)\\
0 & \text{se } (x, y) = (0, 0)
\end{cases}
$$

```python
def func(x, y):
    #if x == 0 and y == 0:
        #return 0
    
    # Saltiamo il punto (0, 0)

    return (x * y) / (x ** 2 + y ** 2)
```

Creiamo la meshgrid:

```python
from matplotlib import cm

lin_x = np.linspace(-5, 5, 200)
lin_y = np.linspace(-5, 5, 200)

x, y = np.meshgrid(lin_x, lin_y)

plt.scatter(x, y, s=1)
plt.scatter(0, 0, facecolors='none', edgecolors='r')

plt.show()
```

```python
ax = plt.axes(projection='3d')

ax.plot_surface(x, y,  func(x, y), cmap = cm.viridis)

# ax.scatter(x, y, f(x, y))

plt.show()
```

Sia $f: \mathbf{R}^2 \rightarrow \mathbf{R}$, definita nel seguente modo:

$$
f(x, y) =
\begin{cases}
{\left( \frac{x^2 y}{x^4 + y^2} \right)}^2 & \text{se } (x, y) \not = (0, 0)\\
0 & \text{se } (x, y) = (0, 0)
\end{cases}
$$

```python
def func(x, y):
    #if x == 0 and y == 0:
        #return 0
    
    # Saltiamo il punto (0, 0)

    return ( ((x**2) * y) / (x ** 4 + y ** 2) ) ** 2
```

```python
from matplotlib import cm

lin_x = np.linspace(-1, 1, 50)
lin_y = np.linspace(-1, 1, 50)

x, y = np.meshgrid(lin_x, lin_y)

plt.scatter(x, y, s=1)
plt.scatter(0, 0, facecolors='none', edgecolors='r')

plt.show()
```

```python
ax = plt.axes(projection='3d')

ax.plot_surface(x, y,  func(x, y), cmap = cm.viridis)

# ax.scatter(x, y, f(x, y))

plt.show()
```
