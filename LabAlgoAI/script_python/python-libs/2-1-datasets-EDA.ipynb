# Data set di esempio

SciKit-Learn mette a disposizione diversi dataset di esempio divisi in tre categorie: 
- toy (giocattolo)
- real word (dataset reali estratti da studi scientifici o altro)
- generati "sinteticamente" con opportuni acccorgimenti


In questa lezione ci occuperemo solo delle prima deue e rimanderemo ad un lezioen successiva la generazione di dataset sintetici.


La categoria **toy** è reperibile al link https://scikit-learn.org/stable/datasets/toy_dataset.html. di questi prenderemo in considerazione:
- **Iris plants dataset** ovvero il dataset degli Iris di Fisher
- **Diabetes dataset**  contenente dati biometrici e anagrafici riguardanti pasìzienti affetti da diabete. Utilizzato per la regressione prendendo come variabile target una misura quantitativa della progressione della malattia.
- **Wine recognition dataset** contenten misurazioni chimiche sulla cresita dell'uva, appartentne a tre diverse cultivar, destinata alla vinicoltura in una regione italiana. 
- **Breast cancer wisconsin (diagnostic) datase** che contiene delle caratteristiche fisiche estratte da immagini per la diagnostica del tummore al seno utiilizzate per diversificare la diagnosi tra tumori benigni e maligni


La categoria **real word** è reperibile al link https://scikit-learn.org/stable/datasets/real_world.html e sono per lo più dataset con classi multiple in cui è necessairo un corposno lavoro di preprocessamento dei dati, specie quelli riguardanti al Natural Language Processing (NLP). Utilizzeremo
- **Forest covertypes** utilizzati per predirre il tipo di species/foresta principale (7 tipi in totale) in quadrati di 30x30 metri utilizzando dati cartografici
- **RCV! dataset** che consiste in una collezione di newswire (articoli di notiziario) memorizzata secondo un datawarehoase in 3 assi tale che per ogni articolo registramo la sua rilevanza su 103 diverse categorie. Le feature sono estratti tramite metodo bag of words.

Da segnalare anche la collezione della University of California, Irvine (UCI), reperibile al link https://archive.ics.uci.edu/ml/index.php .

```python
winedf.describe()
```

```python
%matplotlib inline
import seaborn 
seaborn.boxplot(data=winedf)
```

```python
import numpy as np

scalers = dict()

for c in set(winedf.columns):
    print(c, max(winedf[c]))
    x = winedf[c].to_numpy().reshape(-1, 1)
    scaler = preprocessing.StandardScaler().fit(x)
    winedf[c] = scaler.transform(x)
    scalers[c] = scaler
    
%matplotlib inline
import seaborn 
seaborn.boxplot(data=winedf)
```
